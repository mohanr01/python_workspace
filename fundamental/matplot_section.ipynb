## matplot lib 
1. matplot library is helps for visualization 
2. helps for plot the values

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np
```

```python
x=np.arange(0,10)
y=np.arange(11,21)
```

```python
plt.scatter(x,y,c='g')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph in 2D')
plt.show()
```

```python
plt.plot(x,y,'r-*', label='line 1', linewidth=1)
plt.xlabel('xxxaxis')
plt.ylabel('yyyaxis')
plt.show()
```

```python
#creating subplot
plt.subplot(2,2,1)
plt.plot(x,y,'r-*', label='line 1', linewidth=1)
plt.xlabel('xxxaxis')
plt.ylabel('yyyaxis')
#plt.show()
plt.subplot(2,2,2)
plt.plot(x,y,'g-*', label='line 1', linewidth=1)
plt.xlabel('xxxaxis')
plt.ylabel('yyyaxis')
#plt.show()
plt.subplot(2,2,3)
plt.plot(x,y,'b-*', label='line 1', linewidth=1)
plt.xlabel('xxxaxis')
plt.ylabel('yyyaxis')
#plt.show()
plt.subplot(2,2,4)
plt.plot(x,y,'b-*', label='line 1', linewidth=1)
plt.xlabel('xxxaxis')
plt.ylabel('yyyaxis')
plt.show()
```

```python
sizes = [215, 130, 245, 210]
type(sizes)
```

```python
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0.4, 0, 0, 0)  # explode 1st slice

# Plot
plt.pie(sizes, explode=explode, labels=labels, colors=colors,
autopct='%1.1f%%', shadow=False)

plt.axis('equal')
plt.show()
```
